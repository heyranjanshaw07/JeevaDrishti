```python
import torch

# Use GPU if available
if torch.cuda.is_available():
    device = torch.device("cuda")
else:
    device = torch.device("cpu")
```

```python
from pathlib import Path
from dataloaders.vis_prompt import FewShotDataset

root = Path("/work/mech-ai-scratch/shreyang/FSOD/sampled_dataset/dataset_53_v1")
ds = FewShotDataset(root, source="RatC6")

print("Classes:", ds.get_classes())
ref_img, support_crops = ds.sample_episode("Spindle Cells", n_shots=3)
```

```python
import cv2
import requests
from PIL import Image
from matplotlib import rcParams
import matplotlib.pyplot as plt

# Set figure size
%matplotlib inline
rcParams['figure.figsize'] = 11 ,8

# Input image
image = ref_img
target_sizes = torch.Tensor([image.size[::-1]])

# Query image
query_imgs = support_crops

# Display input image and query image
fig, ax = plt.subplots(1, len(query_imgs) + 1, figsize=(14, 4))
ax[0].imshow(image)
for i, q in enumerate(query_imgs, start=1):
    ax[i].imshow(q)
for a in ax: a.axis("off")
plt.show()
```

```python
from transformers import OwlViTProcessor, OwlViTForObjectDetection

model = OwlViTForObjectDetection.from_pretrained("google/owlvit-base-patch32")
processor = OwlViTProcessor.from_pretrained("google/owlvit-base-patch32")
# Set model in evaluation mode
model = model.to(device)
model.eval()
```

```python
import torch.nn.functional as F

# Pre‐tokenise the three query images → (Nq, 3, H, W)
q_inputs  = processor(images=query_imgs, return_tensors="pt")
q_pixels  = q_inputs["pixel_values"]          # shape = (Nq, 3, 800, 800)

# Average *across* the Nq dimension → (1, 3, H, W)
avg_query_pixels = q_pixels.mean(dim=0, keepdim=True).to(device)

print("avg_query_pixels shape:", avg_query_pixels.shape)  # (1, 3, 800, 800)
```

```python
t_inputs     = processor(images=image, return_tensors="pt").to(device)
pixel_values = t_inputs["pixel_values"]                # (1, 3, H, W)

with torch.no_grad():
    outputs = model.image_guided_detection(
        pixel_values       = pixel_values,
        query_pixel_values = avg_query_pixels,         # ← fused reference
    )

# ----------------- post-process -----------------
target_sizes = torch.tensor([image.size[::-1]])   # (H, W)
results = processor.post_process_image_guided_detection(
    outputs       = outputs,
    threshold     = 0.30,
    nms_threshold = 0.5,
    target_sizes  = target_sizes,
)

boxes, scores = results[0]["boxes"], results[0]["scores"]
print(f"Detections: {len(boxes)} (≥ threshold)")
```

```python
import numpy as np


img = cv2.cvtColor(np.array(image), cv2.COLOR_BGR2RGB)
outputs.logits = outputs.logits.cpu()
outputs.target_pred_boxes = outputs.target_pred_boxes.cpu()

results = processor.post_process_image_guided_detection(outputs=outputs, threshold=0.15, nms_threshold=0.2, target_sizes=target_sizes)
boxes, scores = results[0]["boxes"], results[0]["scores"]

# Draw predicted bounding boxes
for box, score in zip(boxes, scores):
    box = [int(i) for i in box.tolist()]

    img = cv2.rectangle(img, box[:2], box[2:], (255,0,0), 5)
    if box[3] + 25 > 768:
        y = box[3] - 10
    else:
        y = box[3] + 25

plt.imshow(img[:,:,::-1])
```
